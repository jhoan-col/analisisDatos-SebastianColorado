# Análisis Exploratorio de Datos - Dataset del Titanic (Versión 2 en Español)
Este notebook amplía el análisis con medidas estadísticas avanzadas y visualizaciones completamente en español.

```python
# Celda 1: Importar librerías y cargar el Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

df = pd.read_csv('Titanic Dataset.csv')
display(df.head())
```

```python
# Celda 2: Exploración general y valores nulos
print("--- INFORMACIÓN GENERAL ---")
df.info()

print("\n--- VALORES NULOS POR COLUMNA ---")
print(df.isnull().sum())
```

```python
# Celda 3: Medidas Estadísticas Completas (Traducción de nombres para el reporte)
num_cols = ['age', 'fare', 'sibsp', 'parch']
num = df[num_cols].dropna()

resumen = pd.DataFrame({
    "media": num.mean(),
    "mediana": num.median(),
    "moda": num.mode().iloc[0],
    "desv_std": num.std(),
    "varianza": num.var(),
    "rango": num.max() - num.min(),
    "IQR": num.quantile(0.75) - num.quantile(0.25),
    "coef_variacion_%": num.std() / num.mean() * 100,
    "asimetria": num.skew(),
    "curtosis": num.kurt()
})

# Renombrar los índices a español para mayor claridad
resumen = resumen.rename(index={
    'age': 'edad',
    'fare': 'tarifa',
    'sibsp': 'hermanos_conyuges',
    'parch': 'padres_hijos'
})

display(resumen)
```

```python
# Celda 4: Visualización 1 - Diagrama de Cajas y Bigotes (Traducido al español)
plt.figure(figsize=(10, 6))

# Creamos copias traducidas temporalmente para los ejes del gráfico
df_plot = df.copy()
df_plot['clase'] = df_plot['pclass'].map({1: '1ª Clase', 2: '2ª Clase', 3: '3ª Clase'})
df_plot['estado_supervivencia'] = df_plot['survived'].map({0: 'No Sobrevivió', 1: 'Sobrevivió'})

sns.boxplot(x='clase', y='fare', hue='estado_supervivencia', data=df_plot, palette='Set2')
plt.title('Diagrama de Cajas: Tarifa por Clase y Supervivencia', fontsize=14)
plt.xlabel('Clase del Pasajero', fontsize=12)
plt.ylabel('Tarifa Pagada', fontsize=12)
plt.legend(title='Estado')
plt.show()

print('Interpretación: Los pasajeros de 1ª clase pagaron tarifas significativamente más altas con valores atípicos notables, mientras que 3ª clase concentró tarifas bajas.')
```

```python
# Celda 5: Visualización 2 - Mapa de Calor de Correlación (Traducido al español)
plt.figure(figsize=(8, 6))

# Seleccionamos numéricas y renombramos columnas para el mapa de calor
numeric_df = df[['age', 'fare', 'sibsp', 'parch', 'survived', 'pclass']].copy()
numeric_df.columns = ['edad', 'tarifa', 'hermanos_conyuges', 'padres_hijos', 'supervivio', 'clase']
correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Mapa de Calor de Correlación de Variables', fontsize=14)
plt.show()

print('Interpretación: Permite visualizar relaciones lineales entre las variables numéricas del dataset traducidas al español.')
```

```python
# Celda 6: Visualización 3 - Gráfico de Violín (Traducido al español)
plt.figure(figsize=(10, 6))

df_plot2 = df.copy()
df_plot2['genero'] = df_plot2['sex'].map({'male': 'Hombre', 'female': 'Mujer'})
df_plot2['estado_supervivencia'] = df_plot2['survived'].map({0: 'No Sobrevivió', 1: 'Sobrevivió'})

sns.violinplot(x='genero', y='age', hue='estado_supervivencia', data=df_plot2, split=True, palette='muted', inner='quart')
plt.title('Distribución de Edades por Género y Supervivencia', fontsize=14)
plt.xlabel('Género', fontsize=12)
plt.ylabel('Edad (Años)', fontsize=12)
plt.legend(title='Estado')
plt.show()

print('Interpretación: Muestra la densidad de edad por género. Se observa mayor tasa de supervivencia en niños pequeños de ambos sexos y mayor mortalidad en hombres jóvenes.')
```
